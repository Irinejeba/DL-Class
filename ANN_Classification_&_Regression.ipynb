# **ANN Regression -House Prediction**

```python
#ANN Regression -House Prediction
#Import Libraries
import numpy as np
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense,Input
```

```python
#Create input data
x=np.array([
    [1000,2],
    [1500,3],
    [2000,3],
    [2500,4],
    [3000,4],
])
#Create output data
y=np.array([20,30,40,50,55])
```

```python
#Create ANN Model
model=Sequential()
#Input layer
model.add(Input(shape=(2,)))
#Hidden layer 1
model.add(Dense(16,activation='relu'))
#Hidden layer 1
model.add(Dense(8,activation='relu'))
#Output layer
model.add(Dense(1,activation='linear'))
```

```python
#Compile the model
model.compile(optimizer='adam',loss='mse',metrics=['mae'])
```

```python
#Train the moel
model.fit(x,y,epochs=100,verbose=0)
#Make prediction
prediction = model.predict(np.array([[2100,3]]))
print("Predicted House Price:", prediction)
```

# **ANN Binary Classification**


```python
#ANN-Binary Classification -Study Hours
#Step 1: Import libraries
import numpy as np
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense,Input
```

```python
#Create Input layer
x=np.array([
    [1,40],
    [2,50],
    [1,60],
    [1,70],
    [1,80],
    [1,90],
    [1,95]
])

#Create output data
y=np.array([0,0,0,0,1,1,1])
```

```python
#Compile ANN model
model=Sequential()
```

```python
#Input Layer
model.add(Input(shape=(2,)))
#Hidden Layer 1
model.add(Dense(16,activation='relu'))
#Hidden Layer 2
model.add(Dense(8,activation='relu'))
#Output Layer
model.add(Dense(1,activation='sigmoid'))
```

```python
#Compile the model
model.compile(optimizer='adam',loss='binary_crossentropy',metrics=['accuracy'])
```

```python
#Train the model
model.fit(x,y,epochs=100,verbose=0)
#Make prediction
prediction = model.predict(np.array([5,85]))
print("Predicted Probability:", prediction)

#Convert probability into class
if prediction[0][0] >=0.5:
  print("Predicted class:1")
else:
  print("Predicted class:1")
```
